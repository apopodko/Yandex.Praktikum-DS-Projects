Загрузим данные и проведем разведочный анализ

```python
# зададим функцию для импорта данных и вывода первичной информации по ним
# data_name - имя датафрейма для создания, filepath - путь к файлу в формате string
def import_info(dataset_path, sep=',', decimal='.'):
    
    try:
        data_name = pd.read_csv(dataset_path, sep=sep, decimal=decimal, parse_dates=True)
    except:
        return 'Не получилось открыть файл, проверьте имя файла и путь'
    data_name.info()
    display(data_name.head())
    print(f'\nПолных явных дубликатов - {data_name.duplicated().sum()}')
    print('\nid с одинаковыми признаками - '
          f'{data_name.set_index("id").duplicated().sum()}')    
    print(f'\nУникальных id - {data_name["id"].nunique()}')

    return data_name
```

## Исследовательский анализ

Исследуем распределения признаков, посмотрим на наличие выбросов и адекватность

```python
sns.set(font_scale=1.2)
sns.set_style('whitegrid', {'axes.edgecolor': 'black'})

# Определим функцию для графического и описательного анализа столбцов
def describe_column(data, hue=None, alpha=1, ax=None):
    
    # Условие на тип диаграммы не категориальной без сравнения:
    
    if (data.dtype != object) & (hue is None):
        
        fig = plt.figure(figsize=(20,6))
            
        # Настройки сетки графика
        gs = gridspec.GridSpec(2, 2, height_ratios=[5, 1])
        gs.update(hspace=0)
        ax_hist = plt.subplot(gs[0, 0])
        ax_box = plt.subplot(gs[1, 0])
        ax_qq = plt.subplot(gs[:, 1])

        # Boxplot и гистограмма
        sns.boxplot(x=data, ax=ax_box)
        sns.histplot(x=data, 
                     alpha=alpha, 
                     edgecolor='black', 
                     ax=ax_hist,
                     stat='probability')
        plt.subplots_adjust(hspace=0)

        #ax_hist.set_ylabel('Частота', size=14)
        ax_hist.set_title(f'Распределение признака {data.name}', size=14)
        
        # QQ-plot
        sm.qqplot(data, line='s', ax=ax_qq)
        ax_qq.set_title(f'Нормальный график Q-Q признака {data.name}', size=14)
        ax_qq.set_xlabel('Нормальные квантили', size=14)
        ax_qq.set_ylabel('Признаковые квантили', size=14)
        
        plt.show()
        
#        print('Тест Колмогорова-Смирнова на нормальность, p-value - ', 
#              kstest(data, 'norm'))

        # Описательная статистика
        display(pd.DataFrame(data.describe()).T)
```

```python
set_names = [
    'Регион 1',
    'Регион 2',
    'Регион 3'
]

# функция для мультифреймовых графиков
def multiframe(name):

    dataframes = [
        data_0[name], 
        data_1[name], 
        data_2[name]
    ]
    
    df_num = pd.DataFrame()
    for i in range(len(dataframes)):
        df_ = pd.DataFrame(dataframes[i].copy())
        df_['Область'] = set_names[i]
        df_num = pd.concat([df_num, df_])

    fig = plt.figure(figsize=(16,6))
    sns.histplot(x=df_num[name], 
                 hue=df_num['Область'], 
                 edgecolor='black',
                 multiple='layer',
                 common_norm=False, 
                 stat='probability',
                 discrete=None,
                 shrink=1
                )
    
    plt.title(f'Распределение признака {name} в зависимости от региона', size=14)
```

## Обучение и проверка модели

- Будем использовать линейную регрессию, так как остальные алгоритмы менее предсказуемы
- В качестве скейлера будем использовать StandardScaler, больших выбросов нет
- Добавим SimpleImputer на всякий случай, хоть в наших данных нет пропусков, но они могут быть в тестовых

### Подготовка данных и обучение модели

```python
def prepare_train_model(data):
    
    # Делим данные на целевой признак и входные признаки
    X = data.drop('product', axis=1)
    y = data['product']

    # Разделяем данные на тренировочную и валидационную выборки
    X_train, X_valid, y_train, y_valid = train_test_split(
        X,
        y,
        test_size=0.25,
        random_state=RANDOM_STATE
    )
    
    # Создаем пайплайн SimpleImputer+StandardScaler+LinearRegression
    pipe_oil = Pipeline([
        (
            'imputer_before',
             SimpleImputer(missing_values=np.nan, strategy='median')
         ),
        (
            'scaler',
            StandardScaler()
        ),
        (
            'model',
            LinearRegression()
        )

    ])
    
    pipe_oil.fit(X_train, y_train)
    predictions = pipe_oil.predict(X_valid)
    
    df = pd.concat([
        y_valid.reset_index(), 
        pd.Series(predictions, name='predicted_product')], axis=1).drop(['index'], axis=1)
    return df, pipe_oil
```

### Анализ метрик моделей

```python
def model_metrics(y_test, predictions, reg_num):
    
    print(f'Метрики модели для {reg_num} региона')
    print('-----------------------------------------------------------')
    # расчет метрик линейной регрессии
    rmse = root_mean_squared_error(y_test, predictions)
    residuals = y_test - predictions
    r2 = r2_score(y_test, predictions)
    
    print(f'RMSE модели - {rmse.round(2)}')
    print(f'Коэффициент детерминации R2 модели - {round(r2, 2)}')
    print(f'Средние истинные запасы нефти - {y_test.mean().round(2)}')
    print(f'Средние предсказанные запасы нефти - {predictions.mean().round(2)}')
    
    fig, axes = plt.subplots(1, 3, figsize=(18,6))  
    
    # Строим гистограмму
    axes[0].hist(residuals, bins=20, zorder=3, edgecolor='black') 
    axes[0].grid()
    axes[0].set_title('Гистограмма распределения остатков')
    axes[0].set_ylabel('Частота', size=12)
    axes[0].set_xlabel('Остатки', size=12)

    # Построим ящик с усами
    sns.residplot(x=predictions, y=residuals, ax=axes[1], 
                            lowess=True, line_kws=dict(color="r"))
    axes[1].grid()
    axes[1].set_title('Анализ дисперсии')
    axes[1].set_ylabel('Остатки', size=12)
    axes[1].set_xlabel('Предсказания запасов нефти', size=12)
    
    # Сравним распределения запасов истинные и предсказанные
    y_test.hist(bins=20, ax=axes[2], alpha=.5, zorder=3, edgecolor='black')
    predictions.hist(bins=20, ax=axes[2], alpha=.5, zorder=3, edgecolor='black')
    axes[2].grid()
    axes[2].set_title('Распределения запасов сырья')
    axes[2].set_ylabel('Частота', size=12)
    axes[2].set_xlabel('Запасы сырья, тыс. баррелей', size=12)
    axes[2].legend(['Истинные', 'Предсказанные'])
    
    plt.tight_layout()
```

## Подготовка к расчёту прибыли

```python
# определение ключевых значений по ТЗ
BUDGET = 10**10 # бюджет
WELLS_TO_SEARCH = 500 # точек для исследования в регионе
WELLS_TO_DRILL = 200 # точек для разработки
WELL_INCOME = 450000 # доход с каждой единицы продукта (тыс. баррелей)
LOSS_PROBABILITY = 0.025 # порог вероятности убытков
```

```python
# Бюджет на разработку скважины
BUDGET_PER_WELL = BUDGET / WELLS_TO_DRILL

# Достаточный объем сырья для безубыточной разработки
LOSS_FREE_VOLUME = BUDGET_PER_WELL / WELL_INCOME

print(f'Для безубыточной разработки одной скважины необходимо {round(LOSS_FREE_VOLUME, 2)} тыс. баррелей')
print('Нехватка средних объемов до безубыточности в 1 регионе - ' \
    f'{round((1 - data_product_0["product"].mean()/LOSS_FREE_VOLUME) * 100, 2)}%')
print('Нехватка средних объемов до безубыточности в 2 регионе - ' \
    f'{round((1 - data_product_1["product"].mean()/LOSS_FREE_VOLUME) * 100, 2)}%')
print('Нехватка средних объемов до безубыточности в 3 регионе - ' \
    f'{round((1 - data_product_2["product"].mean()/LOSS_FREE_VOLUME) * 100, 2)}%')
```

## Расчёт прибыли и рисков 

```python
def profit(data):
    data.sort_values(by='predicted_product', ascending=False, inplace=True)
    revenue = data.iloc[:WELLS_TO_DRILL, 0].sum() * WELL_INCOME - BUDGET
    return round(revenue)
```

```python
def bootstrap_target(data, reg_num):
    STATE = np.random.RandomState(12345)
    values=[]
    for i in range(1000):
        subsample = data.sample(n=WELLS_TO_SEARCH, replace=True, random_state=STATE)
        values.append(profit(subsample))
    values = pd.Series(values)
    lower = values.quantile(LOSS_PROBABILITY)
    upper = values.quantile(1 - LOSS_PROBABILITY)
    loss = (values < 0).mean() * 100
    print(f'Статистика по {reg_num} региону')
    print(f'Средняя прибыль: {round(values.mean() / 10**6, 3)} млн. рублей')
    print(f'95% CI: ({round(lower / 10**6, 3)}, {round(upper / 10**6, 3)}) млн. рублей')
    print(f'Вероятность убытков {round(loss, 1)} %')
    return values
```
