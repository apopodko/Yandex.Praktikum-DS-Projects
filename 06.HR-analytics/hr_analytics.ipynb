В этом HR-аналитикам пригодится машинное обучение, с помощью которого получится быстрее и точнее отвечать на вопросы бизнеса.

## Описание проекта

**Заказчик**

Отдел HR-аналитики компании «Работа с заботой»

**Входные данные**

Характеристики сотрудников компании. Среди них — уровень удовлетворённости сотрудника работой в компании. Эту информацию получили из форм обратной связи: сотрудники заполняют тест-опросник, и по его результатам рассчитывается доля их удовлетворённости от 0 до 1, где 0 — совершенно неудовлетворён, 1 — полностью удовлетворён

**Задача**

Построить модель, которая сможет предсказать уровень удовлетворённости сотрудника на основе данных заказчика. Построить модель, которая сможет на основе данных заказчика предсказать то, что сотрудник уволится из компании. Сделать выводы и описать предложения для бизнеса


**Цель**

Предсказание оттока — одна из важнейших задач HR-аналитиков. Удовлетворённость работой напрямую влияет на отток сотрудников.  Внезапные увольнения несут в себе риски для компании, особенно если уходит важный сотрудник. HR-отдел пользуясь предсказанием оттока сможет предпринять превентивные меры

**Описание входных данных**
- `id` — уникальный идентификатор сотрудника
- `dept` — отдел, в котором работает сотрудник
- `level` — уровень занимаемой должности
- `workload` — уровень загруженности сотрудника
- `employment_years` — длительность работы в компании (в годах)
- `last_year_promo` — показывает, было ли повышение за последний год
- `last_year_violations` — показывает, нарушал ли сотрудник трудовой договор за последний год
- `supervisor_evaluation` — оценка качества работы сотрудника, которую дал руководитель
- `salary` — ежемесячная зарплата сотрудника
- `job_satisfaction_rate` — уровень удовлетворённости сотрудника работой в компании, целевой признак
- `quit` — увольнение сотрудника из компании

**Прогнозирование уровня удовлетворенности сотрудника**
- `train_job_satisfaction_rate.csv` - тренировочная выборка
- `test_features.csv` - входные признаки тестовой выборки
- `test_target_job_satisfaction_rate.csv` - целевой признак тестовой выборки

**Прогнозирование вероятности увольнения сотрудника**
- `train_quit.csv` - тренировочная выборка
- `test_features.csv` - входные признаки тестовой выборки

```python
sns.set(font_scale=1.2)
sns.set_style('whitegrid', {'axes.edgecolor': 'black'})

# Определим функцию для графического и описательного анализа столбцов
def describe_column(data, hue=None, alpha=1, ax=None):
    
    # Условие на тип диаграммы не категориальной без сравнения:
    
    if (data.dtype != object) & (hue is None):
        
        fig = plt.figure(figsize=(20,6))
        # проверка на непрерывность
        if data.nunique() > 20:
            
            # Настройки сетки графика
            gs = gridspec.GridSpec(2, 2, height_ratios=[5, 1])
            gs.update(hspace=0)
            ax_hist = plt.subplot(gs[0, 0])
            ax_box = plt.subplot(gs[1, 0])
            ax_qq = plt.subplot(gs[:, 1])

            # Boxplot и гистограмма
            sns.boxplot(x=data, ax=ax_box)
            sns.histplot(x=data, 
                         alpha=alpha, 
                         edgecolor='black', 
                         ax=ax_hist,
                         stat='probability')
            plt.subplots_adjust(hspace=0)
            
        # дискретный
        else:
            # Настройки сетки графика
            gs = gridspec.GridSpec(1, 2)
            ax_hist = plt.subplot(gs[0])
            ax_qq = plt.subplot(gs[1])
            
            sns.countplot(x=data, 
                alpha=alpha, 
                edgecolor='black',
                stat = 'probability',
                ax=ax_hist)

        #ax_hist.set_ylabel('Частота', size=14)
        ax_hist.set_title(f'Распределение признака {data.name}', size=14)
        
        # QQ-plot
        sm.qqplot(data, line='s', ax=ax_qq)
        ax_qq.set_title(f'Нормальный график Q-Q признака {data.name}', size=14)
        ax_qq.set_xlabel('Нормальные квантили', size=14)
        ax_qq.set_ylabel('Признаковые квантили', size=14)
        
        plt.show()

        # Описательная статистика
        display(pd.DataFrame(data.describe()).T)
    
    # Условие на тип диаграммы не категориальной со сравнением по признаку:
    elif hue is not None:
        fig = plt.figure(figsize=(12,6))
        
        if data.nunique() < 12:
            discr = True
            shr = 0.8
            multi = 'dodge'
        else:
            discr = None
            shr = 1
            multi = 'layer'  
            
        sns.histplot(x=data, 
                     alpha=alpha, 
                     hue=hue, 
                     edgecolor='black', 
                     common_norm=False, 
                     stat='probability',
                     discrete=discr,
                     shrink=shr,
                     multiple=multi)
        plt.title(f'Распределение признака {data.name} в зависимости от {hue.name}',
                  size=14)
        plt.show()
   
    # Условие на категориальный тип диаграммы
    else:
        if ax is not None:
            data.value_counts().plot(kind='pie', 
                                     autopct=make_autopct(data.value_counts()), 
                                     fontsize=12, ax=ax)
        else:
            data.value_counts().plot(kind='pie', 
                                     autopct=make_autopct(data.value_counts()), 
                                     fontsize=12, figsize=(6, 6))            
        plt.title(f'Распределение признака {data.name}', fontsize=12)
        plt.ylabel('')

# создадим дополнтильную функцию для вывода количества элементов в блоке для графика pie  
# возвращает подпись элемента в % и число
def make_autopct(values):
    def my_autopct(pct):
        total = sum(values)
        val = int(round(pct*total/100.0))
        return '{p:.2f}%\n({v:d})'.format(p=pct,v=val)
    return my_autopct
```

```python
set_names = [
    'Тренировочная выборка 1',
    'Тренировочная выборка 2',
    'Тестовая выборка'
]

# функция для мультифреймовых графиков
def multiframe(name):

    dataframes = [
        data_1_train[name], 
        data_2_train[name], 
        data_test[name]
    ]
    
    # условие на круговой тип диаграммы
    if dataframes[0].dtype == object:
        
        fig, axes = plt.subplots(1, 3, figsize=(16, 6))
        fig.suptitle(f'Распределение признака {name}', fontsize=12)
        for i in range(len(dataframes)):
            describe_column(dataframes[i], ax=axes[i])
            axes[i].set_title(set_names[i], fontsize=12)
            axes[i].set_ylabel('')
        plt.tight_layout()
    
    # условие на числовой тип диаграммы
    else: 
        df_num = pd.DataFrame()
        for i in range(len(dataframes)):
            df_ = pd.DataFrame(dataframes[i].copy())
            df_['Выборка'] = set_names[i]
            df_num = pd.concat([df_num, df_])
        # дискретные  
        if df_num[name].nunique() < 12:
            discr = True
            shr = 0.8
            multi = 'dodge'
        # непрерывные
        else:
            discr = None
            shr = 1
            multi = 'layer'
            
        fig = plt.figure(figsize=(16,6))
        sns.histplot(x=df_num[name], 
                     hue=df_num['Выборка'], 
                     edgecolor='black',
                     multiple=multi,
                     common_norm=False, 
                     stat='probability',
                     discrete=discr,
                     shrink=shr
                    )
        plt.title(f'Распределение признака {name} в зависимости от выборки', size=14)
```

    - `criterion` - `squared_error` или `absolute_error`


- `SVR`
    - `C` - [0.01, 0.25, 0.49, 1.0]
    - `kernel` - [`linear`, `rbf`, `poly`]
    - `degree` - от 2 до 8
    - `gamma` возьмем 0.5, 1.0, `auto` и `scale`
 

- `LGBMRegressor`
    - `boosting_type` - `gbdt`
    - `learning_rate` - 0.05, 0.1 или 0.2
    - `max_depth` - от 4 до 7
    - `n_estimators` - от 100 до 500 с шагом 100
    - `num_leaves` - от 20 до 200 с шагом 20
    - `colsample_bytree` - 0.5, 0.75 или 1.0

### Подготовка данных и оптимизатора поиска

```python
# создадим обучающую и тестовую выборки
X_1_train = data_1_train.drop('job_satisfaction_rate', axis=1)
y_1_train = data_1_train['job_satisfaction_rate']

# списки признаков по типу
ord_cols = [  
    'level',
    'workload'
]

cat_cols = [
    'dept',
    'last_year_promo',
    'last_year_violations'
]

num_cols = [
    'employment_years',
    'supervisor_evaluation',
    'salary'
]
```

Будем искать лучшую модель через Optuna с TPESampler

- В качестве целевой функции будем использовать SMAPE
- Критерий успеха: SMAPE ≤ 15 на тестовой выборке

```python
# Задаем кодировщики
# Encoder = OHE или WOE
def instantiate_encoder(trial : Trial):
    method = trial.suggest_categorical('encoding_method', ['onehot', 'ordinal'])

    if method == 'onehot':
        encoder = instantiate_onehot_encoder(trial)
    elif method == 'ordinal':
        encoder = instantiate_cat_ordinal_encoder(trial)    
        
    return encoder

# OneHotEncoder
def instantiate_onehot_encoder(trial : Trial):
    params = {
        'drop' : 'first',
        'sparse_output' : False
    }
    return OneHotEncoder(**params)


# OrdinalEncoder
def instantiate_cat_ordinal_encoder(trial : Trial):
    params = {
        'handle_unknown' : 'use_encoded_value',
        'unknown_value' : np.nan
    }
    return OrdinalEncoder(**params)

# OrdinalEncoder
def instantiate_ordinal_encoder(trial : Trial):
    params = {
        'handle_unknown' : 'use_encoded_value',
        'unknown_value' : np.nan,
        'categories' : [
            ['junior', 'middle', 'senior'],
            ['low', 'medium', 'high']
        ]
    }
    return OrdinalEncoder(**params)

# Categorical SimpleImputer
def instantiate_categorical_simple_imputer_1(trial : Trial):
    params = {
        'missing_values' : np.nan,
        'strategy' : 'most_frequent'
    }
    return SimpleImputer(**params)

def instantiate_categorical_simple_imputer_2(trial : Trial):
    params = {
        'missing_values' : ' ',
        'strategy' : 'most_frequent'
    }
    return SimpleImputer(**params)

# CAT Pipeline 
# SimpleImputer + Encoder
def instantiate_categorical_pipeline(trial : Trial): 
    pipeline = Pipeline(
        [
            (
                'imputer_1_before', 
                instantiate_categorical_simple_imputer_1(trial)
            ),
            
            (
                'imputer_2_before', 
                instantiate_categorical_simple_imputer_2(trial)
            ),            
            
            (
                'encoder', 
                instantiate_encoder(trial)
            )
        ]
    )
    return pipeline

# ORD Pipeline
# SimpleImputer + OrdinalEncoder + SimpleImputer
def instantiate_ordinal_pipeline(trial : Trial):    
    pipeline = Pipeline(
        [
            (
                'imputer_1_before', 
                instantiate_categorical_simple_imputer_1(trial)
            ),
            
            (
                'imputer_2_before', 
                instantiate_categorical_simple_imputer_2(trial)
            ),

            (
                'ord',
                instantiate_ordinal_encoder(trial)
            ),

            (
                'imputer_after', 
                instantiate_categorical_simple_imputer_1(trial)
            )
        ]
    )
    return pipeline
```

```python
# Задаем скейелеры
# Scaler = Standard or MinMax or Robust 
def instantiate_scaler(trial : Trial):
    method = trial.suggest_categorical(
    'scaling_method', ['standard', 'minmax', 'robust'])
    
    if method == 'standard':
        scaler = instantiate_standard_scaler(trial)
    elif method == 'minmax':
        scaler = instantiate_minmax_scaler(trial)
    elif method == 'robust':
        scaler = instantiate_robust_scaler(trial)
        
    return scaler

# StandardScaler
def instantiate_standard_scaler(trial : Trial):
    return StandardScaler()

# MinMaxScaler
def instantiate_minmax_scaler(trial : Trial) :
    return MinMaxScaler()

# RobustScaler
def instantiate_robust_scaler(trial : Trial):
    return RobustScaler()

# Numerical SimpleImputer
def instantiate_numerical_simple_imputer(trial : Trial):
    params = {
        'missing_values' : np.nan,
        'strategy' : 'median'
    }
    return SimpleImputer(**params)

# Scaler Pipeline
# SimpleImputer + Scaler
def instantiate_numerical_pipeline(trial : Trial): 
    pipeline = Pipeline(
        [
            (
                'imputer_before', 
                instantiate_numerical_simple_imputer(trial)
            ),
            
            (
                'scaler', 
                instantiate_scaler(trial)
            )
        ]
    )
    return pipeline
```

```python
# Задаем модели и определяем гиперпараметры для оптимизации
# Learner =  LinearRegression or DecisionTreeRegressor or SVR or LightGBM
def instantiate_learner_regression(trial : Trial):
    algorithm = trial.suggest_categorical(
    'algorithm', ['elasticnet', 'decisiontree_reg', 'svr', 'lgbm_reg']
    )
    if algorithm == 'elasticnet':
        model = instantiate_elasticnet(trial)
    elif algorithm == 'decisiontree_reg':
        model = instantiate_decision_tree_regressor(trial)
    elif algorithm == 'svr':
        model = instantiate_svr(trial)
    elif algorithm == 'lgbm_reg':
        model = instantiate_lgmb_regressor(trial)      
    return model

# Linear Regression
def instantiate_elasticnet(trial : Trial):
    params = {
        'alpha' : trial.suggest_float('alpha', 0.1, 1.0),
        'l1_ratio' : trial.suggest_float('l1_ratio', 0.0, 1.0),
        'max_iter' : 3000,
        'random_state': RANDOM_STATE
    }
    return ElasticNet(**params)

def instantiate_decision_tree_regressor(trial : Trial):
    params = {
        'max_depth' : trial.suggest_int('max_depth', 1, 10),
        'max_leaf_nodes' : trial.suggest_int('max_leaf_nodes', 20, 200, step=20),
        'min_samples_split' : trial.suggest_int('min_samples_split', 2, 20),
        'criterion': trial.suggest_categorical(
            'criterion', ['squared_error', 'absolute_error']
        ),
        'random_state': RANDOM_STATE
    }
    return DecisionTreeRegressor(**params)

def instantiate_svr(trial : Trial):
    params = {
        'C' :  trial.suggest_float('C', 0.01, 1.01),
        'kernel' : trial.suggest_categorical('kernel', ['linear', 'rbf']),
        'gamma' : trial.suggest_categorical('gamma', ['auto', 'scale']),
    }
    return SVR(**params)

def instantiate_lgmb_regressor(trial : Trial):
    params = {
        'boosting_type' : 'gbdt',
        'learning_rate' : trial.suggest_categorical('learning_rate', [0.05, 0.1, 0.2]),
        'max_depth' : trial.suggest_int('max_depth', 4, 7),
        'n_estimators' : trial.suggest_int('n_estimators', 100, 500, step=100),
        'num_leaves' : trial.suggest_int('num_leaves', 20, 100, step=20),
        'colsample_bytree' : trial.suggest_float('colsample_bytree', 0.5, 1, step=0.25),
        'random_state': RANDOM_STATE   
    }
    return LGBMRegressor(**params)
```

```python
# Задаем препроцессор
# CAT + ORD + NUM
def instantiate_preprocessor(trial : Trial, 
                             cat_cols : list[str], 
                             ord_cols : list[str], 
                             num_cols : list[str]):   
    
    num_pipe = instantiate_numerical_pipeline(trial)
    cat_pipe = instantiate_categorical_pipeline(trial)
    ord_pipe = instantiate_ordinal_pipeline(trial)
    
    preprocessor = ColumnTransformer(
        [
            ('cat', cat_pipe, cat_cols),
            ('ord', ord_pipe, ord_cols),
            ('num', num_pipe, num_cols)
        ],
        remainder='passthrough'
    )
    
    return preprocessor

# Preprocessor + Models
def instantiate_regression_model(trial : Trial, 
                                 cat_cols : list[str], 
                                 ord_cols : list[str], 
                                 num_cols : list[str]):
    
    preprocessor = instantiate_preprocessor(trial, cat_cols, ord_cols, num_cols)
    learner = instantiate_learner_regression(trial)
    
    model = Pipeline(
        [
            ('preprocessor', preprocessor),
            ('models', learner)
        ]
    )
    
    return model
```

```python
# Функция для расчета SMAPE
def smape_metric(y_true, y_pred):
    metric = (100 / len(y_true)) \
    * np.sum(2 * np.abs(y_true - y_pred) / (np.abs(y_true) + np.abs(y_pred)))
    return metric

# Scorer SMAPE
smape = make_scorer(smape_metric)

# Целевая функция первой задачи
def objective_1(trial : Trial, X : pd.DataFrame, y : pd.Series):   
    model = instantiate_regression_model(trial, cat_cols, ord_cols, num_cols)   
    scores = cross_val_score(model, X, y, scoring=smape, cv=5)
    return np.mean(scores)
```

```python
# создадим обучающую и тестовую выборки
X_2_train = data_2_train.drop('quit', axis=1)
y_2_train = data_2_train['quit']

# списки признаков по типу
ord_cols = [  
    'level',
    'workload'
]

cat_cols = [
    'dept',
    'last_year_promo',
    'last_year_violations'
]

num_cols = [
    'employment_years',
    'supervisor_evaluation',
    'salary',
    'job_satisfaction_rate'
]

# кодируем целевую переменную
original_classes = ['no', 'yes']
label_encoder = LabelEncoder()
label_encoder.fit_transform(np.array(original_classes))
label_encoder.classes_ = np.array(original_classes)
y_2_train = label_encoder.transform(y_2_train)
y_2_test = label_encoder.transform(data_2_test_target['quit'])


pd.DataFrame(list(range(len(label_encoder.classes_))), 
             label_encoder.classes_, columns=['LabelEncoder'])
```

Закодировали целевую переменную, `yes` - 1

Будем искать лучшую модель через Optuna с TPESampler

- В качестве целевой функции будем использовать ROC-AUC
- Критерий успеха: ROC-AUC ≥ 0.91 на тестовой выборке

Для подготовки данных будем использовать пайплайны из предыдущей задачи

```python
# Задаем модели и определяем гиперпараметры для оптимизации
# Learner = DecisionTree
def instantiate_classification_learner(trial : Trial):
    
    if trial.params['encoding_method']=='ordinal':
        algorithm = trial.suggest_categorical(
        'algorithm_after_ord', ['decisiontree_class', 'lgbm_class']
        )
        if algorithm == 'decisiontree_class':
            model = instantiate_decision_tree_classifier(trial)
        elif algorithm == 'lgbm_class':
            model = instantiate_lgbm_classifier(trial)
    else:
        algorithm = trial.suggest_categorical(
        'algorithm_after_ohe', ['logreg', 'svc']
        )
        if algorithm == 'logreg':
            model = instantiate_logistic_regression(trial)
        elif algorithm == 'svc':
            model = instantiate_svc(trial)     
    return model

def instantiate_decision_tree_classifier(trial : Trial):
    params = {
        'max_depth' : trial.suggest_int('max_depth', 1, 10),
        'min_samples_leaf' : trial.suggest_int('min_samples_leaf', 1, 40),
        'min_samples_split' : trial.suggest_int('min_samples_split', 2, 20),
        'criterion': trial.suggest_categorical('criterion', ['gini', 'entropy']),
        'random_state': RANDOM_STATE
    }
    return DecisionTreeClassifier(**params)

def instantiate_logistic_regression(trial : Trial):
    params = {
        'solver' : 'saga',
        'penalty' : 'elasticnet',
        'C' : trial.suggest_categorical('C_log', [0.1, 1.0, 10.0]),
        'l1_ratio' : trial.suggest_float('l1_ratio', 0.0, 1.0, step=0.1),
        'max_iter' : 3000,
        'random_state': RANDOM_STATE
    }
    return LogisticRegression(**params)

def instantiate_svc(trial : Trial):
    params = {
        'C' :  trial.suggest_categorical('C_svc', [0.1, 1.0, 10.0]),
        'kernel' : trial.suggest_categorical('kernel', ['linear', 'rbf', 'poly']),
        'degree' : trial.suggest_int('degree', 2, 8),
        'gamma' : trial.suggest_categorical('gamma', ['auto', 'scale']),
        'probability' : True,
        'random_state': RANDOM_STATE
    }
    return SVC(**params)


def instantiate_lgbm_classifier(trial : Trial):
    params = {
        'boosting_type' : 'gbdt',
        'objective' : 'binary',
        'learning_rate' : trial.suggest_categorical('learning_rate', [0.05, 0.1, 0.2]),
        'max_depth' : trial.suggest_int('max_depth', 2, 10),
        'n_estimators' : trial.suggest_int('n_estimators', 100, 500, step=100),
        'num_leaves' : trial.suggest_int('num_leaves', 20, 200, step=20),
        'random_state': RANDOM_STATE   
    }
    return LGBMClassifier(**params)
```

```python
def instantiate_feature_selector(trial : Trial):
    if trial.params['encoding_method']=='ordinal':
        params = {
            'k' : trial.suggest_int('k', 7, 9)
        }
    else:
        params = {
            'k' : trial.suggest_int('k', 7, 11, step=2)
        }        
    return  SelectKBest(**params)

# Preprocessor + Models
def instantiate_classification_model(trial : Trial, 
                                     cat_cols : list[str], 
                                     ord_cols : list[str], 
                                     num_cols : list[str]):
    
    preprocessor = instantiate_preprocessor(trial, cat_cols, ord_cols, num_cols)
    selector = instantiate_feature_selector(trial)
    learner = instantiate_classification_learner(trial)
    
    model = Pipeline(
        [
            ('preprocessor', preprocessor),
            ('selector', selector),
            ('models', learner)
        ]
    )
    
    return model
```

```python
# Задаем оценку для поиска из roc_auc_score и recall_score    
def objective_2(trial : Trial, X : pd.DataFrame, y : pd.Series):   
    model = instantiate_classification_model(trial, cat_cols, ord_cols, num_cols)
    roc_auc_scorer = make_scorer(roc_auc_score, needs_proba=True)
    scores = cross_val_score(model, X, y, scoring=roc_auc_scorer, cv=5)
    return np.mean(scores)
```

```python
tpe_study_2 = create_study(
    study_name = 'HR-analitcs_classification',
    direction= 'maximize',
    sampler=TPESampler(seed=RANDOM_STATE)
)

tpe_study_2.optimize(lambda trial: objective_2(trial, X_2_train, y_2_train), 
                   n_trials=100, show_progress_bar=True)
```

### Метрики модели

Посмотрим на другие метрики лучшей модели

```python
# Создадим функцию для вывода метрики модели
# Принимает на вход y_test - тестовая выборка, predictions - предсказание целевого признака
def log_model_metrics(y_test, predictions, probabilities, name):
    print('ROC-AUC score -', round(roc_auc_score(y_test, probabilities), 3))
    print(f'Accuracy - {round(accuracy_score(y_test, predictions), 3)}')
    print(f'Recall - {round(recall_score(y_test, predictions), 3)}')
    print(f'Precision - {round(precision_score(y_test, predictions), 3)}')
    cm = confusion_matrix(y_test, predictions)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues_r')
    plt.title(f'Матрица ошибок {name}')
    plt.xlabel('Предсказанный класс')
    plt.ylabel('Истинный класс')
```

- обратить внимание на мидл и сеньор уровень сотрудников по модели и для первых - рассмотреть перечисленные выше рекомендации
- основная текучка происходит в джуниор позициях, можно отслеживать ситуацию с помощью моделей в целом или конкретных джунов, которые подают надежды
- при необходимости для влиять на них через выравнивание загруженности, зарплату или повышением в зависимости от ситуации, так как с увеличением продолжительности работы в компании сильно падает вероятность увольнения
- для общей ситуации обратить внимание у всех рисковых сотрудников на уровень загрузки, он не должен быть слишком мал или велик
- для заблоговременного реагирования можно опираться на оценку руководителя, как впрочем обычно и делают, и предпринимать шаги заранее
