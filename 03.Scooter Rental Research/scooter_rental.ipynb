## Описание проекта
[(к содержанию)](#Содержание)

**Заказчик** - отдел маркетинга сервиса проката самокатов GoFast

**Входные данные** - информация о пользователях, их поездках и подписках из мобильного приложения GoFast

**Задача** - анализ поездок и проверка гипотез

**Цель** - выявить гипотезы, которые могут оказать положительное влияние на развитие бизнеса GoFast

**Оригинальные данные** 
- `/datasets/users_go.csv` - данные о пользователях сервиса
- `/datasets/rides_go.csv` - данные о поездках
- `/datasets/subscriptions_go.csv` - данные о подписках

**Описание данных**
- Таблица `users_go`
    - `user_id` - уникальный идентификатор пользователя
    - `name` - имя пользователя
    - `age` - возраст
    - `city` - город
    - `subscription_type` - тип подписки (free, ultra)


- Таблица `rides_go`
    - `user_id` - уникальный идентификатор пользователя
    - `distance` - расстояние, которое пользователь проехал в текущей сессии (в метрах)
    - `duration` - продолжительность сессии (в минутах) — время с того момента, как пользователь нажал кнопку «Начать поездку» до момента, как он нажал кнопку «Завершить поездку»
    - `date` - дата совершения поездки


- Таблица `subscriptions_go`
    - `subscription_type` - тип подписки
    - `minute_price` - стоимость одной минуты поездки по данной подписке
    - `start_ride_price` - стоимость начала поездки
    - `start_ride_price` - стоимость ежемесячного платежа

Проверим, что айдишники такие же, как и в первой таблице. Т.к. в таблице `data_users` айди это натуральные числа от 1 до 1534 без пропусков, то сумма уникальных айди в `data_rides` должна быть равна n*(n+1)/2, где n=1534, тогда у нас железно не будет случая, где вместо `user_id` = 5 будет 1583 к примеру

```python
data_rides['user_id'].unique().sum() ==(
    data_rides['user_id'].nunique() * (data_rides['user_id'].nunique() + 1) / 2)
#(data_rides['user_id'].unique() == data_users['user_id']).sum() # второй вариант без математики
```

Такое расстояние проезжают обычно за 20-25 минут, возможно проблемы с геолокацией, либо с таймером поездки. Посмотрим на корреляцию продолжительности и расстояния за поездку

```python
# построим распределение продолжительности поездок от пройденного расстояния
data_rides.plot(kind='scatter', x='distance', y='duration', c=None, grid=True, \
                zorder=2, figsize=(18,6))
plt.title('Распределение продолжительности поездок от пройденного расстояния', fontsize=14)
plt.xlabel('Расстояние за поездку, м', fontsize=12)
plt.ylabel('Продолжительность поездки, минуты', fontsize=12);
```

Видим, что есть неплохая линейная зависимость, которая для нашей правки вполне подойдет, ведь всего 95 выбивающихся значений (справа внизу четко видны). Попробуем найти коэффициенты из линейной регрессии

```python
# создадим модель линейной регрессии, поменяем размерность столбца с расстояниями,
# чтобы матрично все решалось
model = LinearRegression()
x_ = data_rides.query('duration > 0.5')['distance'].values.reshape(-1,1)
y_ = data_rides.query('duration > 0.5')['duration']
model.fit(x_, y_)
model.score(x_, y_)
```

Коэф. детерминации 0.34, в приницпе неплохо, для нашей задачи подходит. Построим прямую регрессии на распределении

```python
# построим распределение продолжительности поездок
# от пройденного расстояния и добавим аппроксимационную прямую регрессии
data_rides.plot(kind='scatter', x='distance', y='duration', c=None, grid=True, \
                zorder=2, figsize=(18,6))
plt.axline((0, model.intercept_), slope=model.coef_, color='r')
plt.title('Распределение продолжительности поездок от пройденного расстояния', fontsize=14)
plt.xlabel('Расстояние за поездку, м', fontsize=12)
plt.ylabel('Продолжительность поездки, минуты', fontsize=12);
```

Выглядит хорошо, так как выбивающихся значений мало, на статистику они окажут меньше влияния, если мы поправим их по регрессии, чем если просто выбросим. Поправим время в этих поездках

```python
data_rides.loc[data_rides['duration']==0.5,'duration'] =(
    data_rides.loc[data_rides['duration']==0.5, 'distance'] * model.coef_ + model.intercept_)
```

Отлично, аномалия исправлена, распределение близко к нормальному. - **Нулевая гипотеза H0** - пользователи с подпиской тратят столько же времени на поездку, что и пользователи без подписки
- **Альтернативная гипотеза H1** - пользователи с подпиской тратят больше времени на поездку, чем пользователи без подписки

Она правосторонняя, кроме того, помним факт о разнице дисперсий выборок

### Дополнительные задачи по распределениям

#### Промокоды

**Задача**

Отделу маркетинга GoFast поставили задачу: нужно провести акцию с раздачей промокодов на один бесплатный месяц подписки, в рамках которой как минимум 100 существующих клиентов должны продлить эту подписку. То есть по завершении периода действия подписки пользователь может либо отказаться от неё, либо продлить, совершив соответствующий платёж.

Эта акция уже проводилась ранее и по итогу выяснилось, что после бесплатного пробного периода подписку продлевают 10 % пользователей. Выясните, какое минимальное количество промокодов нужно разослать, чтобы вероятность не выполнить план была примерно 5 %. Подберите параметры распределения, описывающего эту ситуацию, постройте график распределения и сформулируйте ответ на вопрос о количестве промокодов.

**Решение**

Стартовое количество промокодов возьмем за 1000, так как нам необходимо привлечь 100 подписчиков, а вероятность подписки - 10 %. Можно было бы воспользоваться биномиальным распределением и считать вручную, но с такими параметрами (**n=1000** и **p=0.1**) мы можем спокойно его аппроксимировать нормальным распределением. Для пущей уверенности проверим на 3 отклонения. Минимально необходимое количество промокодов будет больше 1000 на 10-20 %, попробуем его найти

```python
# зададим начальные параметры биномиального распределения
n = 1000
p = 0.1

# посчитаем сразу среднее и ст.оклонение для аппрокисмации нормальным распределением

mu = n * p
sigma = sqrt(n * p * (1 - p))

# проверяем условие применимости аппроксимации
if (mu - (3 * sigma) >= 0) and ((mu + (3 * sigma) <= n)):
    print('Условие аппроксимации выполняется')
    
    # рассчитаем количество промокодов, при которых вероятность не выполнить план будет около 5%
    prob = st.norm(mu, sigma).cdf(100)
    while (prob > 0.05):
        n += 1
        mu = n * p
        sigma = sqrt(n * p * (1 - p))
        prob = st.norm(mu, sigma).cdf(100)
    
    print(f'Необходимо разослать {n} промокодов, тогда вероятность не выполнить план - {prob.round(4)}')

    # зададим границы для графика, полные нам не нужны
    left = int(mu - 5 * sigma)
    right = int(n * p + 5 * sigma)
    x = np.linspace(left, right, 100)
    x1 = np.linspace(left, right, 500)
    # построим графики, чтобы убедиться, что наша аппроксимация была оправдана
    fig, ax = plt.subplots(figsize=(18,6))
    ax.plot(x, st.binom.pmf(x, n, p), 'bo', ms=4, label='Вероятности биномиального распределения')
    ax.plot(x1, st.norm.pdf(x1, mu, sigma), 'r-', lw=3, alpha=0.3, \
            label='Плотность вероятности нормального распределения')
    plt.axvline(x=st.norm(mu, sigma).ppf(0.05), color='r')
    plt.fill_between(x1, y1=st.norm.pdf(x1, mu, sigma), where = x1 < st.norm(mu, sigma).ppf(0.05), \
                     color='r', alpha=0.2, label='Вероятность не выполнить план =< 0.05')
    ax.legend(loc='best', fontsize=12)
    ax.set_xlabel('Количество промокодов', fontsize=12)
    plt.title(f'Распределение вероятностей выполнения плана при {n} разосланных промокодах')
    plt.yticks(fontsize=12)

else:
    print('Условие аппроксимации не выполняется')
```

Для того, чтобы вероятность не выполнить план была примерно 5 %, при условиях поставленной задачи, нам необходимо разослать минимум 1169 промокодов, тогда вероятность не выполнить план будет 4.97 %. Так как мы использовали аппроксимацию, и необходимо иметь запас, то можно рекомендовать разослать 1180 - 1200 промокодов, для большей уверенности, что план будет выполнен. Проверим через биномиальное распределение

```python
n = 1000
prob = 1

while (prob > 0.05):
    n += 1 
    prob = st.binom.cdf(100, n, p)
    
print(f'Необходимо разослать {n} промокодов, тогда вероятность не выполнить план - {prob.round(4)}')
```

По биномальному распределению для выполнения задачи необходим разослать минимум 1172 промокода, тогд вероятность невыполнения плана будет 4.95 %. Наша рекомендация подошла, разница между аппроксимацией и обычным подходом составила 3 промокода, но даже здесь для большей надежности наша рекомендация сохраняется

#### push-уведомления
[(к содержанию)](#Содержание)

**Задача**

Отдел маркетинга рассылает клиентам push-уведомления в мобильном приложении. Клиенты могут открыть его или не открывать. Известно, что уведомления открывают около 40 % получивших клиентов. Отдел планирует разослать 1 млн уведомлений. С помощью аппроксимации постройте примерный график распределения и оцените вероятность того, что уведомление откроют не более 399,5 тыс. пользователей.

**Решение**

Применим подход с аппроксимацией из задачи выше еще раз, так как здесь уже слишком много "опытов", n - 10^6

```python
# зададим начальные параметры биномиального распределения
n = 1000000
p = 0.4

# посчитаем сразу среднее и ст.оклонение для аппрокисмации нормальным распределением

mu = n * p
sigma = sqrt(n * p * (1 - p))
   
# рассчитаем количество промокодов, при которых вероятность не выполнить план будет около 5%
prob = st.norm(mu, sigma).cdf(399500)
    
print(f'При отправке {n} push-уведомлений, вероятность, что их откроют',
      'не более 399500 пользователей - {prob.round(4)}')

# зададим границы для графика, полные нам не нужны
left = int(mu - 5 * sigma)
right = int(n * p + 5 * sigma)
x = np.linspace(left, right, 100)
x1 = np.linspace(left, right, 500)
# построим графики
fig, ax = plt.subplots(figsize=(18,6))
ax.plot(x, st.binom.pmf(x, n, p), 'bo', ms=4, label='Вероятности биномиального распределения')
ax.plot(x1, st.norm.pdf(x1, mu, sigma), 'r-', lw=3, alpha=0.3, \
        label='Плотность вероятности нормального распределения')
plt.axvline(x=399500, color='r')
plt.xticks(list(plt.xticks()[0]) + [399500])
plt.fill_between(x1, y1=st.norm.pdf(x1, mu, sigma), where = x1 < st.norm(mu, sigma).ppf(0.05), \
                 color='r', alpha=0.2, label='Вероятность открытия уведомлений =< 0.05')

ax.legend(loc='best', fontsize=12)
ax.set_xlabel('Количество push-уведомлений', fontsize=12)
plt.title('Распределение вероятностей открытия уведомлений 1 млн пользователей')
plt.yticks(fontsize=12);
```

Не более 399500 пользователей из 1 млн. откроют push-уведомления с вероятностью 15.37 %
