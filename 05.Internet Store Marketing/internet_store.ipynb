# Персонализация предложений интернет-магазина "В один клик"

Интернет-магазин «В один клик» продаёт разные товары: для детей, для дома, мелкую бытовую технику, косметику и даже продукты. Отчёт магазина за прошлый период показал, что активность покупателей начала снижаться. Привлекать новых клиентов уже не так эффективно: о магазине и так знает большая часть целевой аудитории. Возможный выход — удерживать активность постоянных клиентов. Сделать это можно с помощью персонализированных предложений.

## Описание проекта

**Заказчик** 

Интернет-магазин "В один клик"

**Входные данные** 
- поведение покупателя на сайте
- коммуникация с покупателем
- продуктовое поведении покупателя
- выручка, которую получает магазин с покупателя
- время, которое покупатель провёл на сайте в течение периода
- среднемесячная прибыль от покупателя за последние 3 месяца

**Задача**

- необходимо промаркировать уровень финансовой активности постоянных покупателей
- построить модель, которая предскажет вероятность снижения покупательской активности клиента в следующие три месяца
- выделить сегменты покупателей и разработать для них персонализированные предложения

**Цель**

Разработать решение, которое позволит персонализировать предложения постоянным клиентам, чтобы увеличить их покупательскую активность

**Описание входных данных**

- **`market_file.csv`** - данные о поведении покупателя на сайте, о коммуникациях с покупателем и его продуктовом поведении
    - `id` - номер покупателя в корпоративной базе данных
    - `Покупательская активность` - рассчитанный класс покупательской активности (**целевой признак**): `снизилась` или `прежний уровень`
    - `Тип сервиса` -  уровень сервиса, например `премиум` и `стандарт`
    - `Разрешить сообщать` - информация о том, можно ли присылать покупателю дополнительные предложения о товаре
    - `Маркет_актив_6_мес` - среднемесячное значение маркетинговых коммуникаций на клиента за последние 6 месяцев
    - `Маркет_актив_тек_мес` - количество маркетинговых коммуникаций в текущем месяце
    - `Длительность` - количество дней прошедших с момента регистрации покупателя на сайте
    - `Акционные_покупки` - среднемесячная доля покупок по акции от общего числа покупок за последние 6 месяцев
    - `Популярная_категория` - самая популярная категория товаров у покупателя за последние 6 месяцев
    - `Средний_просмотр_категорий_за_визит` - среднее количество категорий, просмотренных покупателем за визит за последний месяц
    - `Неоплаченные_продукты_штук_квартал` - общее число неоплаченных товаров в корзине за последние 3 месяца
    - `Ошибка_сервиса` - число сбоев, которые коснулись покупателя во время посещения сайта
    - `Страниц_за_визит` - среднее количество страниц, которые просмотрел покупатель за один визит на сайт за последние 3 месяца


- **`market_money.csv`** - данные о выручке, которую получает магазин с покупателя
    - `id` - номер покупателя в корпоративной базе данных
    - `Период` - название периода, во время которого зафиксирована выручка. Например, `текущий_месяц` или `предыдущий_месяц`
    - `Выручка` - сумма выручки за период


- **`market_time.csv`** - данные о времени, которое покупатель провёл на сайте в течение периода
    - `id` - номер покупателя в корпоративной базе данных
    - `Период` - название периода, во время которого зафиксировано общее время
    - `минут` — значение времени, проведённого на сайте, в минутах


- **`money.csv`** -  данные о среднемесячной прибыли на покупателя за последние 3 месяца
    - `id` - номер покупателя в корпоративной базе данных
    - `Прибыль` - значение прибыли

## Исследовательский анализ

```python
# Определим функцию для графического и описательного анализа столбцов
def describe_column(data, hue=None, alpha=1):
    
    # Условие на тип диаграммы не категориальной:
    
    if (data.dtype != object) & (hue is None):
        # Настройки сетки графика
        fig = plt.figure(figsize=(20,6))
        gs = gridspec.GridSpec(2, 2, height_ratios=[5, 1])
        gs.update(hspace=0)
        ax_hist = plt.subplot(gs[0, 0])
        ax_box = plt.subplot(gs[1, 0])
        ax_qq = plt.subplot(gs[:, 1])
        
        # Boxplot и гистограмма
        sns.boxplot(x=data, ax=ax_box)
        sns.histplot(x=data, alpha=alpha, edgecolor='black', ax=ax_hist)
        sns.set(font_scale=1.2)
        sns.set_style('whitegrid', {'axes.edgecolor': 'black'})
        ax_hist.set_ylabel('Частота', size=14)
        ax_hist.set_title(f'Распределение признака {data.name}', size=14)
        plt.subplots_adjust(hspace=0)
        
        # QQ-plot
        sm.qqplot(data, line='s', ax=ax_qq)
        ax_qq.set_title(f'Нормальный график Q-Q признака {data.name}', size=14)
        ax_qq.set_xlabel('Нормальные квантили', size=14)
        ax_qq.set_ylabel('Признаковые квантили', size=14)
        
        plt.show()

        # Описательная статистика
        display(pd.DataFrame(data.describe()).T)
    
    elif hue is not None:
        fig = plt.figure(figsize=(12,6))
        sns.histplot(x=data, alpha=alpha, hue=hue, edgecolor='black', common_norm=False, stat='probability')
        plt.title(f'Распределение признака {data.name} в зависимости от {hue.name}', size=14)
        plt.show()
   
    # Условие на категориальный тип диаграммы
    else:
        data.value_counts().plot(kind='pie', autopct=make_autopct(data.value_counts()), 
                                 figsize=(6,6), fontsize=12)
        plt.title(f'Распределение признака {data.name}', fontsize=12)
        plt.ylabel('')
        plt.show()

# создадим дополнтильную функцию для вывода количества элементов в блоке для графика pie  
# возвращает подпись элемента в % и число
def make_autopct(values):
    def my_autopct(pct):
        total = sum(values)
        val = int(round(pct*total/100.0))
        return '{p:.2f}%\n({v:d})'.format(p=pct,v=val)
    return my_autopct
```

### Объединение таблиц

В фреймах `data_money` и `data_time` собранные данные по месяцам, надо сделать сводные таблицы по пользователям, чтобы потом объединить в один фрейм

```python
# сводная таблица по пользователям из data_money
data_money_ = pd.pivot_table(
        data_money, 
        values='выручка', 
        index='id', 
        columns='период').rename_axis(columns=None)
data_money_.columns = 'выручка_' + data_money_.columns
data_money_.reset_index(inplace=True)

# сводная таблица по пользователям из data_time
data_time_ = pd.pivot_table(
        data_time, 
        values='минут', 
        index='id', 
        columns='период').rename_axis(columns=None)
data_time_.columns = 'минут_' + data_time_.columns
data_time_.reset_index(inplace=True)

print('Размерность data_customers - ', data_customers.shape)
print('Размерность data_money_ - ', data_money_.shape)
print('Размерность data_time_ - ', data_time_.shape)
display(data_money_.head(1))
display(data_time_.head(1))
```

Все в порядке, теперь объединим все фреймы в один

```python
# список из всех фреймов
data_frames = [data_customers, data_money_, data_time_]

# применим reduce для списка фреймов
data = reduce(lambda left, right: pd.merge(left, right, on=['id'], how='inner'), data_frames)

print(data.shape)
data.head()
```

```python
# создадим обучающую и тестовую выборки
X = data_ml.drop('покупательская_активность', axis=1)
y = data_ml['покупательская_активность']
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    stratify=y,
    random_state=RANDOM_STATE  
)

# кодируем целевую переменную
original_classes = ['Прежний уровень', 'Снизилась']
label_encoder = LabelEncoder()
label_encoder.fit_transform(np.array(original_classes))
label_encoder.classes_ = np.array(original_classes)
y_train = label_encoder.transform(y_train)
y_test = label_encoder.transform(y_test)


# списки признаков по типу
ord_cols = [  
    'акционные_покупки_класт',
    'тип_сервиса',
    'разрешить_сообщать'
]

ohe_cols = [
    'популярная_категория'
]

num_cols = [
    'маркет_актив_6_мес',
    'длительность',
    'выручка_темп',
    'минут_сред',
    'страниц_за_визит',
    'средний_просмотр_категорий_за_визит',
    'неоплаченные_продукты_штук_квартал',
    'ошибка_сервиса',
    'маркет_актив_тек_мес'
]

pd.DataFrame(list(range(len(label_encoder.classes_))), 
             label_encoder.classes_, columns=['LabelEncoder'])
```

Будем искать лучшую модель через Optuna с TPESampler

- В качестве целевой функции булем использовать roc_auc_score и recall_score. Recall важен, так как заказчик хочет определить клиентов со сниженной активностью и удержать их, значит надо в первую очередь найти их максимальное число и разослать предложения
- Кроме того, для экономии бюджета, нужно минимизовать число рассылок для пользователей без снижения активности

```python
# Задаем кодировщики и пайплан для них для ord колонок
def instantiate_encoder(trial : Trial):
    method = trial.suggest_categorical(
      'encoding_method', ['ordinal', 'onehot']
    )

    if method=='ordinal':
        encoder = instantiate_ordinal_encoder(trial)
    elif method=='onehot':
        encoder = instantiate_onehot_encoder(trial)

    return encoder

def instantiate_ordinal_encoder(trial : Trial):
    return OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=np.nan)

def instantiate_onehot_encoder(trial : Trial):
    return OneHotEncoder(drop='first', sparse_output=False)

def instantiate_categorical_pipeline(trial : Trial):
    pipeline = Pipeline(
        [
            ('imputer_before', SimpleImputer(missing_values=np.nan, strategy='most_frequent')), 
            ('encoder', instantiate_encoder(trial)),
            ('imputer_after', SimpleImputer(missing_values=np.nan, strategy='most_frequent'))
        ]
    )
    return pipeline

# SimpleImputer + OHE
ohe_pipe = Pipeline(
    [
        (
            'simpleImputer_ohe', 
            SimpleImputer(missing_values=np.nan, strategy='most_frequent')
        ),
        
        (
            'ohe', 
            OneHotEncoder(drop='first', sparse_output=False)
        )
    ]
)
```

```python
# Задаем скейелеры
def instantiate_scaler(trial : Trial):
    method = trial.suggest_categorical(
    'scaling_method', ['standard', 'minmax', 'robust']
    )
    if method == 'standard':
        scaler = instantiate_standard_scaler(trial)
    elif method == 'minmax':
        scaler = instantiate_minmax_scaler(trial)
    elif method == 'robust':
        scaler = instantiate_robust_scaler(trial)
        
    return scaler

def instantiate_standard_scaler(trial : Trial):
    return StandardScaler()

def instantiate_minmax_scaler(trial : Trial) :
    return MinMaxScaler()

def instantiate_robust_scaler(trial : Trial):
    return RobustScaler()
```

- `decision_tree_classifier` - лучший набор гиперпараметров определим по статье https://arxiv.org/abs/1812.02207
    - `max_depth` количество гиперпараметров - 1
    - `min_samples_leaf` max=40
    - `min_samples_split` max=20
    
    
- `logistic_regression`
    - будем использовать только `elasticnet` в связке с `saga` и `l1_ratio`, так могут быть и с 0 - l2 или 1 - l1 регуляризация
    - `C` возьмем пул логарифмических значений на 1 порядок от 1 в обе стороны
    - Повысим максмальное число итераций, чтобы считалось до конца
    
- `svc`
    - `C` возьмем пул логарифмических значений на 1 порядок от 1 в обе стороны
    - будем рассматривать три ядра
    - `gamma` возьмем 0.1 и 1.0
    
- `KNN`
    - от 51 до 501 соседа, не больше, так как у нас будет использована кроссвалидация и в выборке может быть меньше записей
    - все метрики по расстоянию

```python
# задаем модели и определяем гиперпараметры для оптимизации
def instantiate_learner(trial : Trial):
    algorithm = trial.suggest_categorical(
    'algorithm', ['decisiontree', 'logreg', 'svc', 'knn']
    )
    if algorithm == 'decisiontree':
        model = instantiate_decision_tree_classifier(trial)
    elif algorithm == 'logreg':
        model = instantiate_logistic_regression(trial)
    elif algorithm == 'svc':
        model = instantiate_svc(trial)
    elif algorithm == 'knn':
        model = instantiate_knn(trial)      
    return model

def instantiate_decision_tree_classifier(trial : Trial):
    params = {
        'max_depth' : trial.suggest_int('max_depth', 1, 10),
        'min_samples_leaf' : trial.suggest_int('min_samples_leaf', 1, 40),
        'min_samples_split' : trial.suggest_int('min_samples_split', 2, 20),
        'criterion': trial.suggest_categorical('criterion', ['gini', 'entropy']),
        'random_state': RANDOM_STATE
    }
    return DecisionTreeClassifier(**params)

def instantiate_logistic_regression(trial : Trial):
    params = {
        'solver' : 'saga',
        'penalty' : 'elasticnet',
        'C' : trial.suggest_categorical('C_log', [0.1, 1.0, 10.0]),
        'l1_ratio' : trial.suggest_float('l1_ratio', 0.0, 1.0, step=0.1),
        'max_iter' : 3000,
        'random_state': RANDOM_STATE
    }
    return LogisticRegression(**params)

def instantiate_svc(trial : Trial):
    params = {
        'C' :  trial.suggest_categorical('C_svc', [0.1, 1.0, 10.0]),
        'kernel' : trial.suggest_categorical('kernel', ['linear', 'rbf']),
        'degree' : trial.suggest_int('degree', 2, 8),
#        'gamma' : trial.suggest_categorical('gamma', 'sacle'),
#        'probability' : True,
        'random_state': RANDOM_STATE
    }
    return SVC(**params)

def instantiate_knn(trial : Trial):
    params = {
        'n_neighbors' : trial.suggest_int('n_neighbors', 51, 501, step=50),
        'weights' : trial.suggest_categorical('weights', ['uniform', 'distance']),
        'metric' : trial.suggest_categorical('metric', ['minkowski', 'euclidean', 'manhattan'])        
    }
    return KNeighborsClassifier(**params)
```

```python
# Задаем препроцессор, для OHE - обычный пайплайн без перебора
def instantiate_preprocessor(trial : Trial, ohe_cols : list[str], ord_cols : list[str], num_cols : list[str]):
    
    scaler = instantiate_scaler(trial)
    encoder_pipe = instantiate_categorical_pipeline(trial)
    preprocessor = ColumnTransformer([
        ('ohe', ohe_pipe, ohe_cols),
        ('ord', encoder_pipe, ord_cols),
        ('num', scaler, num_cols)],
        remainder='passthrough'
    ) 
    return preprocessor

# Общий пайплайн
def instantiate_model(trial : Trial, ohe_cols : list[str], ord_cols : list[str], num_cols : list[str]):
    
    preprocessor = instantiate_preprocessor(trial, ohe_cols, ord_cols, num_cols)
    learner = instantiate_learner(trial)   
    model = Pipeline(
    [
        ('preprocessor', preprocessor),
        ('models', learner)
    ])
    return model
    
# Задаем оценку для поиска из roc_auc_score и recall_score    
def objective(trial : Trial, X : pd.DataFrame, y : pd.Series):   
    model = instantiate_model(trial, ohe_cols, ord_cols, num_cols)
    scores = cross_validate(model, X, y, scoring=('roc_auc', 'recall'), cv=5)
    mean_auc_score = np.mean(scores['test_roc_auc'])
    mean_rec_scores = np.mean(scores['test_recall'])
    return mean_auc_score, mean_rec_scores
```

```python
tpe_study = create_study(
    study_name = 'Optuna_four_models_proj_6',
    directions= ['maximize', 'maximize'],
    sampler=TPESampler(seed=RANDOM_STATE)
)

tpe_study.optimize(lambda trial: objective(trial, X_train, y_train), 
                   n_trials=200, show_progress_bar=True)
```

### Выбор лучшей модели

Теперь выберем лучшую модель из топ-20 обучающей выборки и топ-5 на тестовой по обеим целевым функциям

```python
# эти колонки отбросим
cols_ = [
    'number', 
    'datetime_start', 
    'datetime_complete', 
    'duration', 
    'state'
]

# выгрузим фрейм всех попыток и почистим от дубликатов
best_models = (
    tpe_study.trials_dataframe()
    .drop(cols_, axis=1)
    .drop_duplicates()
    .drop_duplicates(subset=['values_0', 'values_1'])
)

# доп столбец среднее из двух показателей и оставим 30 лучших моделей
best_models.insert(2, 'mean_val', np.mean(best_models[['values_0', 'values_1']], axis=1))
best_models = best_models.nlargest(30, 'mean_val')

# считаем показатели на тестовой выборке
for trial_num in best_models.index:
    model = instantiate_model(tpe_study.trials[trial_num], ohe_cols, ord_cols, num_cols)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    best_models.loc[trial_num, 'values_0_test']=(roc_auc_score(y_test, y_pred))
    best_models.loc[trial_num, 'values_1_test']=(recall_score(y_test, y_pred))
    
# оставим из 10 лучших по тестовой выборке, 5 лучших по обучающей
best_models.insert(3, 'mean_val_test', 
                   np.mean(best_models[['values_0_test', 'values_1_test']], axis=1))
best_models.insert(1, 'values_0_test', best_models.pop('values_0_test'))
best_models.insert(3, 'values_1_test', best_models.pop('values_1_test'))
best_models.nlargest(10, 'mean_val_test').nlargest(5, 'mean_val')
```
