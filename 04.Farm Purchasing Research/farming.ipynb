# Модель для снижения рисков хозяйства при покупке коров

Фермер, владелец молочного хозяйства «Вольный луг», хочет купить бурёнок, чтобы расширить поголовье стада коров. Для этого он заключил выгодный контракт с ассоциацией пастбищ «ЭкоФерма». Условия позволяют фермеру очень тщательно отобрать коров. Он определяет качество молока по строгой методике, и при этом ему нужно выполнять свой план развития молочного хозяйства. 



Фермер хочет, чтобы каждая бурёнка давала не менее 6000 килограммов молока в год, а её надой был вкусным — строго по его критериям.


Необходимо разработать модель машинного обучения, которая поможет ему управлять рисками и принимать объективное решение о покупке. «ЭкоФерма» готова предоставить подробные данные о своих коровах.

## Описание проекта

**Заказчик** - фермер, владелец молочного хозяйства «Вольный луг»

**Входные данные** -  данные о стаде фермера на текущий момент и о коровах «ЭкоФермы», которых фермер хочет изучить перед покупкой

**Задача** - определить целевые параметры и входные признаки в данных о стаде фермера. Создать модель линейной регрессии для прогнозирования удоя. Создать модель логистической регрессии для прогнозирования вкуса молока. Оценить коров «ЭкоФермы» с помощью моделей, дать оценку качества и описать модели

**Цель** - разработать модель, которая поможет ему управлять рисками и принимать объективное решение о покупке


**Описание оригинальных данных**

- **`ferma_main.csv`** - данные о стаде фермера на текущий момент
    - **Характеристики коровы**
        - `id` - уникальный идентификатор коровы
        - `Порода` - порода коровы
        - `порода папы_быка` - порода папы коровы

Создадим функции для вывода описательной статисткики

```python
# Определим функцию для графического и описательного анализа столбцов
# data - столбец данных
# title - название для графика - string
# x_label - подпись оси Х - string
# binsize - размер bin
# возвращает датафрейм с опистальной статистикой столбца
def describe_column(data, title=' ', x_label=' ', binsize='auto'):
    # Условие на тип диаграммы не категориальной:
    if data.dtype != object:
        # Общая структура
        fig, axes = plt.subplots(2, 1, figsize=(12,7), 
                                 gridspec_kw={'height_ratios': [4, 1]})
        plt.suptitle(title, size=14)
    
        # Строим гистограмму
        axes[0].hist(data, bins = binsize, zorder=3, edgecolor='black') 
        axes[0].grid()
        axes[0].tick_params(labelbottom=True)
        axes[0].set_ylabel('Частота', size=12)
        axes[0].set_xticklabels([])

        # Построим ящик с усами
        axes[1].boxplot(data, vert=False)
        axes[1].grid()
        axes[1].set_yticks([])
        axes[1].set_xlabel(x_label, size=12)
        plt.tight_layout()
        plt.subplots_adjust(hspace=0)

        # Описательная статистика
        return pd.DataFrame(data.describe()).T
    
    # Условие на категориальный тип диаграммы
    else:
        data.value_counts().plot(kind='pie', autopct=make_autopct, 
                                 figsize=(12,6), fontsize=12)
        plt.title(title, fontsize=14)
        plt.ylabel('')

# создадим дополнтильную функцию для вывода количества элементов в блоке для графика pie  
# возвращает подпись элемента в % и число
def make_autopct(values):
    def my_autopct(pct):
        total = sum(values)
        val = int(round(pct*total/100.0))
        return '{p:.2f}%  ({v:d})'.format(p=pct,v=val)
    return my_autopct

# создадим функцию для построения сравнительной диаграммы
# data - датафрейм
# num_param - численный столбец
# cat_param - категорийный столбец
# title - название для графика - string
# x_label - подпись оси Х - string
# show_stat - показать статистику по подгруппам - True или False
def comparison_hist(data, num_param, cat_param=None, title='', x_label='', show_stat=True):    
    # построим диаграмму сравнения
    plt.figure(figsize=(15,6))
    ax = sns.histplot(data=data, x=num_param, hue=cat_param, alpha=0.7, 
                      common_norm=False, stat='probability')
    plt.grid()
    ax.set_axisbelow(True)
    plt.title(title, fontsize=14)
    plt.xlabel(x_label, fontsize=12)
    plt.ylabel('Частота', fontsize=12)

    if show_stat == True:
        # расчитаем статистику по группам
        stat = (data
                .groupby(cat_param)[num_param]
                .describe().reset_index()[[cat_param, 'count', 'mean', 'std']])
        return stat
```

## Прогнозирование удоя коров
[(к содержанию)](#Содержание)

Создадим функции для подготовки данных и вывода метрики

```python
# создадим функцию подготовки данных для линейной регрессии
# все категориальные переменные трансформируются с помощью One-Hot Enconding
# все численные входные признаки масштабируются и нормализуются
# входные данные X - датафрейм входных признаков, y - целевой признак
# возвращает разбивку на тренировочный и тестовый сеты X_train, X_test, y_train, y_test
# OHE модель и модель масштабирования признаков ohe_encoder, std_scaler
def data_preparation(X, y, stratify=None):
    cat_cols = X.select_dtypes(include=['object']).columns.tolist()
    num_cols = X.select_dtypes(exclude=['object']).columns.tolist()

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        random_state=RANDOM_STATE,
        stratify=stratify)

    # One-Hot Encoding
    ohe_encoder = OneHotEncoder(drop='first', sparse=False)

    X_train_ohe = ohe_encoder.fit_transform(X_train[cat_cols])
    X_test_ohe = ohe_encoder.transform(X_test[cat_cols])

    X_train_ohe = pd.DataFrame(X_train_ohe, 
                               columns=ohe_encoder.get_feature_names(cat_cols))
    X_test_ohe = pd.DataFrame(X_test_ohe, 
                              columns=ohe_encoder.get_feature_names(cat_cols))

    # Scaling
    std_scaler = StandardScaler()

    X_train_scaled = std_scaler.fit_transform(X_train[num_cols])
    X_test_scaled = std_scaler.transform(X_test[num_cols])

    X_train_scaled = pd.DataFrame(X_train_scaled, columns=num_cols)
    X_test_scaled = pd.DataFrame(X_test_scaled, columns=num_cols)

    # concat sets
    X_train = pd.concat([X_train_scaled, X_train_ohe], axis=1)
    X_test = pd.concat([X_test_scaled, X_test_ohe], axis=1)
    
    print('Категориальные признаки модели - ', cat_cols)
    print('Численные признаки модели - ', num_cols)
    
    return X_train, X_test, y_train, y_test, ohe_encoder, std_scaler

# создадим функцию вывода метрики R2 и остатков модели
# входные данные y_test - тестовая часть целевого параметра
# predictions - предсказанный моделью целевой параметр
# возвращает residuals, r2, mse, rmse, mae
def model_metrics(y_test, predictions):

    # расчет метрик линейной регрессии
    mse = mean_squared_error(y_test, predictions)
    rmse = mean_squared_error(y_test, predictions, squared=False)
    mae = mean_absolute_error(y_test, predictions)
    residuals = y_test - predictions
    r2 = r2_score(y_test, predictions)

    print(f'Коэффициент детерминации R2 модели - {r2.round(3)}')
    
    fig, axes = plt.subplots(1, 2, figsize=(15,6))  
    # Строим гистограмму
    axes[0].hist(residuals, bins=20, zorder=3, edgecolor='black') 
    axes[0].grid()
    axes[0].set_title('Гистограмма распределения остатков')
    axes[0].set_ylabel('Частота', size=12)
    axes[0].set_xlabel('Остатки', size=12)

    # Построим ящик с усами
    #axes[1].scatter(x=predictions, y=residuals, zorder=3)
    axes[1] = sns.residplot(x=predictions, y=residuals, 
                            lowess=True, line_kws=dict(color="r"))
    axes[1].grid()
    axes[1].set_title('Анализ дисперсии')
    axes[1].set_ylabel('Остатки', size=12)
    axes[1].set_xlabel('Предсказания удоя', size=12)   
    
    return residuals, r2, mse, rmse, mae
```

```python
# Получим вероятности всех предсказаний
probablities = model_clf_1.predict_proba(X_test)

# Найдем все значения precision, recall и порога для тестовой выборки и предсказаний
precision, recall, threshold  = precision_recall_curve(y_test, probablities[:,1])

# Построим графики зависимостей
plt.figure(figsize=(15,6))
plt.title('Зависимость Recall и Precision от порога', fontsize=14)
plt.plot(threshold, precision[:-1], 'b--', label='Precision')
plt.plot(threshold, recall[:-1], 'r--', label='Recall')
plt.xlabel('Порог', fontsize=12)
plt.legend(loc='lower left', fontsize=12)
plt.ylim([0,1.005])
plt.xlim([min(threshold)-0.001,max(threshold)])
plt.grid();
```

Из графиков видим, что Precision достигает 1 при пороге около **0.72-0.74**, при этом Recall не уходит слишком сильно в ноль при таком значении. Варьируя входные параметры модели и зная точную задачу (в терминах метрики - мы будем устремлять Precision к 1), можно использовать этот график как дополнительную метрику модели, чтобы оценивать связку трех параметров - порог, Precision и Recall

Найдем минимальный порог, при котором precision = 1 и посчитаем метрику еще раз

```python
# Получим минимальный порог, при котором precision = 1
min_precision_threshold = threshold[precision[:-1]==1][0]

# Немного варьируем порог и рассчитываем метрику модели с порогом
pred_with_threshold = (probablities[:,1] > min_precision_threshold).astype(int)
print(f'Порог равен {min_precision_threshold}')
log_model_metrics(y_test, pred_with_threshold)
```
