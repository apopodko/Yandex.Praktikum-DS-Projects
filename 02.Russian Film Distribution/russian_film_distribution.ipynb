```python
data_movies.loc[4638,'puNumber'] = 221154311
data_movies.drop(index=5067, inplace=True)
data_movies.reset_index(drop=True)
print('Дубликатов в номере прокатного удостоверения - ', data_movies['puNumber'].duplicated().sum())
```

```python
# Создадим сводную таблицу финансовых показтелей по режиссеру
gov_movies_by_director = (data_gov_movies
 .pivot_table(index='main_director', values=['total_gov_support', 'box_office', 'budget'], aggfunc='sum')
 .reset_index()
                         )
gov_movies_by_director.columns = ['main_director', 'total_box_office', 'total_budget', 'total_gov_support']

#Добавим столбец с средним рейтингом фильмов режиссера
gov_movies_by_director = gov_movies_by_director.merge(data_gov_movies 
                                                       .pivot_table(index='main_director', values='ratings') 
                                                       .reset_index(), on='main_director', how='outer')
```

```python
# Построим график распределения господдержки по режиссерам (топ-50)
ax = (gov_movies_by_director.sort_values('total_gov_support').tail(50)
      .plot('main_director', 'total_gov_support', kind='bar', figsize=(18,9))
     )
plt.ylabel('Размер поддержки', size=12)
plt.legend(['Полная господдержка'])

# Построим график распределения среднего рейтинга по режиссерам
gov_movies_by_director.sort_values('total_gov_support').tail(50).plot('main_director', 'ratings', \
                         secondary_y=True, ax=ax, rot=90, color='red', grid=True, marker='o', \
                         label='Средний рейтинг', xlabel='Режиссер', mark_right=False)
plt.ylabel('Средний рейтинг фильмов у режиссера', size=12)
plt.title('Распределение господдержки по режиссерам (топ-50)');
```

Видно, что режиссеры получающие большую господдержку, не всегда имеют хорошие рейтинги фильмов (Ф.Бондарчук). И наоборот, режиссеры с маленькой господдержкой могут получать серьезное одобрение со стороны зрителей. Посмотрим подробнее на топ-20 режиссеров по господдержке

По сборам эти категории в конце и начале периода примерно равны, что, вероятно, и объясняется разницей в количестве картин в данных категориях. 

    - **4. Фильмы с государственной поддержкой**
        - **Общие зависимости**
            - МинКульт предоставляет только невозвратное финансирование. 16.2 млрд. невозвратной поддержки (80.6%) и 3.9 млрд. возвратной (19.4%). ~14.2 млрд за все время выделил Фонд Кино и 4.8 млрд - МинКульт
            - Господдержка начинает заметно проявляться с 2013-2014 годов, начиная с ~0.5 млрд и выходит на более менее стабильный уровень в 4 млрд в год в 2015 году. При этом доля невозвратных средств составляет большую часть поддержки - 3-3.5 млрд, тогда как возвратных средств - 0.5-1 млрд. в год
            - Бюджет фильмово в 2015 году возрастает до 7 млрд и держится примерно на этом уровне до 2018 года, немного снижаясь, а в 2019 году резко возрастает до ~12.5 млрд. рублей. До 2013 года нет данных, скорее всего не собирались централизовано. Суммарыне сборы больше, чем полная гос.поддержка во все года
            - Лишь в 2017 и 2018 году, полные сборы были больше суммарного бюджета всех фильмов. В первую очередь сборы идут на возврат частных инвестиций, оплата маркетинга и т.д. Однозначно сказать, были ли возвращены деньги государству за фильмы нельзя 

        - **Зависимости поддержки по жанрам**
            - Больше всего было выделено денег на фильмы в жанре драма - ~5.5 млрд. Далее идут комедия (3.2 млрд), мультфильм (2.4 млрд), фантастика (1.4 млрд), фэнтези (1 млрд). Меньше всего на реальное ТВ - < 0.1 млрд. Суммарыне сборы хорошо коррелируют с количеством фильмов в жанрах 
            - Успешнее всего за период 2013-2019 были фильмы в жанре комедия (1.2 млрд. против 0.6 млрд.). Далее идут мультфильмы (0.58 млрд), фантастика (0.21 млрд) и мелодрама (0.17 млрд). Лидер по бюджетной поддержке - драма, не дотягивает до окупаемости (0.95 млрд. против 1.12 млрд.)
            - Лидер по медианной поддержке с большим отрывом - исторический жанр (~250 млн.), за ним идет фантастика - 80 млн. и мультфильмы - 65 млн. Меньше всего выделяют на фильм в жанре мелодрама - 25 млн. В жанре история всего пять фильмов, окупаемость у них 0.75, но сборы больше господдержки. Фильмы в жанре драма и комедия суммарно получают большие сборы за счет количества и присутствия блокбастеров 

        - **Зависимости по рейтингу и режиссерам**
            - Медианный рейтинг фильмов с господдержкой - 6.2 с наблюдамемыми границами [3.2, 8.5]. За период 2013-2019 года средний рейтинг находится примерно на одном значении, около 6 со стандартным отклонением 1 
            - Режиссеры получающие большую господдержку, не всегда имеют хорошие рейтинги фильмов (Ф.Бондарчук). И наоборот, режиссеры с маленькой господдержкой могут получать серьезное одобрение со стороны зрителей
            - Размер господдержки не коррелирует с рейтингом фильмов у режиссеров, как и со сборами у режиссеров. Хотя заметны весьма успешные режиссеры (К.Шипенко или А.Мегердичев) с удачными фильмами тепло принятыми зрителями (Холоп) и большой окупаемостью (коэф. 3-5 к бюджету фильма). Возможно, имеет смысл в пересмотре программы финансирования и отбора фильмов
            - Режиссеры с небольшими сборами (Тодоровский и Учитель) могут иметь большой средний рейтинг (7.3-7.6). Эти и другие режиссеры представляют другую часть кинематографа, где не может быть больших сборов, так как это по большей части авторское или арт-хаусное кино, но тем не менее в рамках культуры оно также имеет большое значение 

        - **Зависимости по типу фильмов**
            - Из всех категорий, поддержку получают только художественные и анимационные фильмы, возможно остальные категории получают поддержку через другие структуры, например МинОбразования, в 2019 году началась поддержка документального кино 
            - Анимационные фильмы получают поддержку примерно на одном уровне с 2015 года, около 0.5 млрд. Основная часть средств идет на поддержку художественных фильмов. Сборы составляют за аналогичный период 1 млрд в год
            - Поддержка художественных фильмов к 2015 году резко возрастает с 0.5 до 3 млрд и далее с локальным минимумом в 17-18 году к 2019 году достигает 4 млрд. рублей. Сборы быстро растут с 2014 года с 0.8 млрд до 9 млрд в 2017 - и далее находится примерно на этом уровне 

        - **Корреляция сборов от господдержки**
            - Возможно, есть небольшая корреляция сборов от господдержки, в логарифмическом масштабе распределение немного похоже на график корня. При начале поддержке государством фильма резко растут сборы фильмы с 10-100 тысяч до 10-100 млн рублей, и при дальнейшем увеличении господдержки сборы выходят на плато со слабым ростом в районе 1 млрд рублей 

- **Рекомендации заказчику**
    - рекомендую посмотреть на дубликаты в номере прокатного удостоверения, так как это уникальный идентификатор и могут быть сбои в базе данных
    - обратить внимание на сбор данных о кассовых сборах, для оценки работы программы поддержки фильмов желательно знать сборы у всех фильмов, если правообладатель не предоставляет данные, и они вообще нигде не учитываются, может плохо сказаться в дальнейшем для успешности и рентабельности господдержки. Вероятно, все же отчетность о поддержке присутствует на других уровнях
    - доля фильмов с информацией о сборах в российских кинотеатрах заметно выросла с 2010 года, и к 2019 году составляет около 60%, что является вполне комфортным уровнем для всех
    - возможно, стоит обратить внимание на поддержку фильмов 6+ и 12+, так как их количество в два раза меньше 16+ и 18+, при близких сборах в 12+ и 16+, есть потенциал для 12+. В категории 0+ очень мало кинолент, теоретически так же есть потенциал, но менее реализуемый из-за узости аудитории
    - в фильмах с господдержкой 2018-2019 году продолжается увеличение поддержки, рост кассовых сборов при этом остановился, необходимо оценить ситуацию, так как до этого был сильный рост сборов
    - очень много лент в жанре драма с господдежкой, при этом суммарная окупаемость меньше единицы, тогда как в жанрах фэнтези и мультфильм хорошая окупаемость, а количество лент заметно меньше, возможно, имеет смысл перераспределить поддержку в другие жанры, более успешные у зрителей
    - вероятно, есть необходимость в пересмотре политики по поддержке режиссеров из топ-10 в пользу менее известных, так как рейтинги и сборы у последних часто выше, чем у первых
    - имеет смысл продолжать поддержку независимого кино, так как оно важно для культуры, рейтинги у некоторых режиссеров очень хорошие
    - вероятно, стоит обратить внимание на документальное кино, поддержки все это время нет, спрос есть
    - возможно, необходимо продолжать наращивать объемы поддержки, так как наблюдается положительная корреляция между сборами и размером господдержки, но при этом развивать отбор фильмов с учетом комментариев
