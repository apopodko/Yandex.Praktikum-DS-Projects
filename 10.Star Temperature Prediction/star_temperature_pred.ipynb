## Разведочный анализ

```python
def import_and_analyze(path, sep=',', decimal='.', date_cols=True, **params):
    try:
        data = pd.read_csv(path, 
                           sep=',', 
                           decimal ='.', 
                           parse_dates=date_cols, 
                           **params)
    except:
        print('Не удалось открыть файл, проверьте путь')
    
    data.info()
    print(f'\nПолных явных дубликатов - {data.duplicated().sum()}')
    display(data.head())
    
    
    
    stats = {'values' : [
                         'Unique, count',
                         'Unique, %',
                         'Missing, count',
                         'Missing, %'
                        ]}
    for col in data:
        stats[col] = [
                      data[col].nunique(),
                      round(100 * data[col].nunique() / data.shape[0], 2),
                      data[col].isnull().sum(),
                      round(100 * data[col].isnull().sum() / data.shape[0], 2)
                     ]
    stats = pd.DataFrame(stats).set_index('values').T
    display(stats.sort_values('Missing, %', ascending=False))
    display(data.describe())
    
    return data
```

### Исследовательский анализ

```python
sns.set(font_scale=1.2)
sns.set_style('whitegrid', {'axes.edgecolor': 'black'})

# Определим функцию для графического и описательного анализа столбцов
def describe_column(data, explode=None):

    # Условие на тип диаграммы не категориальной без сравнения:

    if (data.dtype != object):
        fig = plt.figure(figsize=(20,6))

        # Настройки сетки графика
        gs = gridspec.GridSpec(2, 2, height_ratios=[5, 1])
        gs.update(hspace=0)
        ax_hist = plt.subplot(gs[0, 0])
        ax_box = plt.subplot(gs[1, 0])
        ax_qq = plt.subplot(gs[:, 1])

        # Boxplot и гистограмма
        sns.boxplot(x=data, ax=ax_box)
        sns.histplot(x=data,
                     bins='sturges',
                     alpha=1,
                     edgecolor='black',
                     ax=ax_hist,
                     stat='probability')
        plt.subplots_adjust(hspace=0)
        ax_hist.set_ylabel('Частота', size=14)
        ax_hist.set_title(f'Распределение признака {data.name}', size=14)

        # QQ-plot
        pg.qqplot(data, ax=ax_qq, square=False)
        ax_qq.set_title(f'Нормальный график Q-Q признака {data.name}', size=14)
        ax_qq.set_xlabel('Нормальные квантили', size=14)
        ax_qq.set_ylabel('Признаковые квантили', size=14)

        plt.show()

        # Описательная статистика
        display(pd.DataFrame(data.describe()).T)

    # Условие на категориальный тип диаграммы
    else:
        data.value_counts().plot(kind='pie', explode=explode,
                                 autopct=make_autopct(data.value_counts()),
                                 fontsize=12, figsize=(6, 6))
        plt.title(f'Распределение признака {data.name}', fontsize=12)
        plt.ylabel('')

# создадим дополнтильную функцию для вывода количества элементов в блоке для графика pie
# возвращает подпись элемента в % и число
def make_autopct(values):
    def my_autopct(pct):
        total = sum(values)
        val = int(round(pct*total/100.0))
        return '{p:.2f}%\n({v:d})'.format(p=pct,v=val)
    return my_autopct
```

```python
# OHE
ohe_pipe = Pipeline([
    (
        'imputer_before', 
         SimpleImputer(missing_values=np.nan, strategy='constant', fill_value='unknown')
    ),
    (
        'ohe',
        OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False)
    )
])

# ORD
ord_pipe = Pipeline([
    (
        'imputer_before', 
         SimpleImputer(missing_values=np.nan, strategy='constant', fill_value='other')
    ),
    (
        'ord',
         OrdinalEncoder(categories = [['red', 
                                      'light orange', 
                                      'yellow-white', 
                                      'white', 
                                      'blue-white', 
                                      'blue']],
                        handle_unknown='use_encoded_value', 
                        unknown_value=np.nan)
    ),
    (
        'imputer_after', 
         SimpleImputer(missing_values=np.nan, strategy='most_frequent')
    )    
])

#NUM
num_pipe = Pipeline([
    (
        'num_imputer_before', 
         SimpleImputer(missing_values=np.nan, strategy='mean')
    ),
    (
        'scaler',
        MinMaxScaler()
    )
])

# Preprocessor = CAT + ORD + NUM
preprocessor = ColumnTransformer(
    [
        ('ohe', ohe_pipe, ohe_cols),
        ('ord', ord_pipe, ord_cols),
        ('num', num_pipe, num_cols)
    ],
    remainder='passthrough')
```

## Построение базовой нейронной сети

Создадим класс базовой нейронной сети с 2 скрытыми слоями и инициализацией весов по Каймингу. На ее основе потом будем осуществлять подбор гиперпараметров

```python
class Net_Init(nn.Module):
  
    def __init__(self, n_in_neurons, n_hidden_neurons_1, 
                                     n_hidden_neurons_2,
                                     n_out_neurons,
                                     drop_1,
                                     drop_2,
                                     act_layer_1,
                                     act_layer_2,
                                     batch_norm=False):
        super(Net_Init, self).__init__()
        torch.manual_seed(42)
        self.batch_norm = batch_norm        
        
        self.fc1 = nn.Linear(n_in_neurons, n_hidden_neurons_1)
        self.act1 = act_layer_1
        if self.batch_norm==True:
            self.bn1 = nn.BatchNorm1d(n_hidden_neurons_1)
        self.drop1 = nn.Dropout(drop_1)
        
        self.fc2 = nn.Linear(n_hidden_neurons_1, n_hidden_neurons_2)
        self.act2 = act_layer_2
        if self.batch_norm==True:
            self.bn2 = nn.BatchNorm1d(n_hidden_neurons_2)
        self.drop2 = nn.Dropout(drop_2)

        self.fc3 = nn.Linear(n_hidden_neurons_2, n_out_neurons)
        
        self.apply(self.init_weights)

    def forward(self, x):
        
        x = self.act1(self.fc1(x))
        if self.batch_norm==True:
            x = self.bn1(x)
        x = self.drop1(x)
        
        x = self.act2(self.fc2(x))
        if self.batch_norm==True:
            x = self.bn2(x)
        x = self.drop2(x)
        x = self.fc3(x)
        return x
    
    def init_weights(self, layer):
        if isinstance(layer, nn.Linear):
            nn.init.kaiming_normal_(layer.weight)
            nn.init.normal_(layer.bias, mean=0, std=1)
```

Создадим класс для преобразования входных данных, тренировочных и тестовых, в формат Dataset для последюущего использования в Dataload для батчей

```python
class Net_Dataset(Dataset):
    
    def __init__(self, X, y):
        self.X = torch.FloatTensor(X)
        self.y = torch.FloatTensor(y.values)
        
    def __len__(self):
        return self.X.shape[0]
    
    def __getitem__(self, index):
        return (self.X[index], self.y[index])
```

Создадим класс ранней остановки

```python
class EarlyStopping():
    
    def __init__(self, hold, delta_part):
        self.hold = hold
        self.delta_part = delta_part
        self.counter = 0
        self.min_test_loss = float('inf')
        self.early_stop = False
    
    def __call__(self, test_loss):
        if test_loss < self.min_test_loss:
            self.min_test_loss = test_loss
            self.counter = 0        
        
        elif test_loss > self.min_test_loss * (1 + self.delta_part):
            self.counter += 1
            if self.counter >= self.hold:  
                self.early_stop = True
```

Создадим класс обучения модели

```python
class Net_Trainer():
    """
    Parameters:
        dataset: Класс Net_Dataset, предобрабатывающий данные
        loss_func: Функция потерь
        learning_rate: Скорость обучение (градиентный шаг)
        num_epochs: Полное количество эпох
        batch_size: Размер батча
        early_stopping: Количество эпох без улучшений до остановки обучения
        optim: Оптимизатор
        
    Attributes:
        start_model: Начальная необученная модель
        best_model: Лучшая модель по функции потерь, после обучения
        train_loss: Средние значения функции потерь на тренировочных 
                    данных в каждую эпоху
        test_loss: Средние значения функции потерь на валидационных 
                  данных в каждую эпоху

    Methods:
        fit: Обучение модели
        predict: Возвращает предсказание лучшей обученной модели

    """    
    
    def __init__(self, dataset, net, loss_func=nn.MSELoss, learning_rate=1e-3, 
                 num_epochs=100, batch_size=32, early_stopping=50, 
                 optim=torch.optim.Adam):
        
        self.dataset = dataset
        self.zero_model = net
        self.loss_func = loss_func
        self.learning_rate = learning_rate
        self.num_epochs = num_epochs
        self.batch_size = batch_size
        self.early_stopping = early_stopping
        self.optim = optim
        
        self.train_loss = []
        self.test_loss = []
        self.best_model = net
        torch.manual_seed(42)
    def predict(self, X):
        return self.best_model(torch.FloatTensor(X)).detach().numpy()

    def fit(self, X_train, y_train, X_test, y_test):
        torch.manual_seed(42)
        net = self.zero_model
        
        optimizer = self.optim(net.parameters(), lr=self.learning_rate)
        
        train = self.dataset(X_train, y_train)
        test = self.dataset(X_test, y_test)

        train = DataLoader(train, 
                           batch_size=self.batch_size, 
                           shuffle=False,
                           num_workers=0,
                           worker_init_fn=lambda _: np.random.seed(42)) 
        test = DataLoader(test, 
                          batch_size=self.batch_size, 
                          shuffle=False,
                          num_workers=0,
                          worker_init_fn=lambda _: np.random.seed(42))
        
        best_loss = float('inf') # лучшее значение фнукции потерь
        best_epoch = 0           # эпоха, где достигнут best_loss
        
        early_stopper = EarlyStopping(hold=self.early_stopping, delta_part=0.01) 
        torch.manual_seed(42)
        for epoch in range(self.num_epochs):
        # for epoch in tqdm(range(self.num_epochs), desc='Train...', unit=' epochs'):
            net.train()        
            mean_loss = 0
            batch_n = 0

            for batch_X, target in train:
                optimizer.zero_grad()

                preds = net.forward(batch_X).flatten()
                loss_value = self.loss_func(preds, target)
                loss_value.backward()
                optimizer.step()

                mean_loss += float(loss_value)
                batch_n += 1

            mean_loss /= batch_n
            self.train_loss.append(mean_loss)

            net.eval()
            mean_loss = 0
            batch_n = 0

            with torch.no_grad():
                for batch_X, target in test:
                    preds = net.forward(batch_X).flatten()
                    loss_value = self.loss_func(preds, target)

                    mean_loss += float(loss_value)
                    batch_n += 1

            mean_loss /= batch_n
            self.test_loss.append(mean_loss)
  
            early_stopper(mean_loss)
            if early_stopper.counter == 1:
                self.best_model = net
                best_epoch = epoch
            elif early_stopper.early_stop:
            #    print(f'Остановка обучения. Loss: {round(mean_loss))}')
                break
```

Создадим функцию построения графика обучения модели

```python
def train_report(train_loss, test_loss):
    plt.figure(figsize=(12, 6))
    plt.plot(train_loss, color='green', label='Train loss')
    plt.plot(test_loss, color='blue', label='Test loss')
    plt.title('Визуализация процесса обучения:')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend(loc='lower left')
    plt.show()
```

Инициализируем базовый вариант нейросети со следующими параметрами

```python
base_net_params = {
    'n_in_neurons': X_train.shape[-1],
    'n_hidden_neurons_1': 130,
    'n_hidden_neurons_2': 180,
    'n_out_neurons': 1,
    'drop_1': 0.0,
    'drop_2': 0.0,
    'batch_norm': False,
    'act_layer_1': nn.Tanh(),
    'act_layer_2': nn.LeakyReLU()  
}

base_net = Net_Init(**base_net_params)
```

Инициализируем базовый класс обучения со следующими параметрами и обучим модель

```python
base_train_params = {
    'dataset': Net_Dataset,
    'net': base_net,
    'num_epochs': 4000, 
    'learning_rate': 5e-3,
    'early_stopping': 50,
    'batch_size': 192,
    'loss_func': nn.SmoothL1Loss(),
    'optim': torch.optim.Adam
}

base_model = Net_Trainer(**base_train_params)
base_model.fit(X_train, y_train, X_test, y_test)
```

## Улучшение нейронной сети

Улучшение нейронной сети будем проводить подбором гиперпаматеров с помощью HyperOpt. Будем использовать созданный нами класс.

```python
def objective(params):

    net = Net_Init(params['n_in_neurons'], 
               params['n_hidden_neurons_1'], 
               params['n_hidden_neurons_2'],
               params['n_out_neurons'],
               params['drop_1'],
               params['drop_2'],
               params['act_layer_1'],
               params['act_layer_2'],
               params['batch_norm'])

    model = Net_Trainer(dataset=params['dataset'], 
                      net=net, 
                      optim=params['optim'], 
                      loss_func=params['loss_func'], 
                      num_epochs=params['num_epochs'], 
                      learning_rate=params['learning_rate'], 
                      batch_size=params['batch_size'])
    
    model.fit(X_train, y_train, X_test, y_test)
    y_pred = model.predict(X_test)
    score = round(root_mean_squared_error(y_test, y_pred))
    
    # Return results
    return {'loss': score,
            'status': STATUS_OK,
            'model': model,
            'net': net,
            'params': params
           }
```

```python
act_funcs = [nn.ReLU(), nn.Tanh(), nn.LeakyReLU(), nn.ELU(), nn.LogSigmoid()]
param_space = {
                'n_in_neurons': X_train.shape[-1],
                'n_hidden_neurons_1': hp.choice('n_hidden_neurons_1', 
                                                np.arange(40, 200, 10, dtype=int)),
                'n_hidden_neurons_2': hp.choice('n_hidden_neurons_2', 
                                                np.arange(40, 200, 10, dtype=int)),
                'n_out_neurons': 1,
                'drop_1': hp.choice('drop_1', [0.0, 0.15, 0.3]),
                'drop_2': hp.choice('drop_2', [0.0, 0.15, 0.3]),
                'act_layer_1': hp.choice('act_layer_1', act_funcs),
                'act_layer_2': hp.choice('act_layer_2', act_funcs),
                'batch_norm': hp.choice('batch_norm', [True, False]),
                'dataset': Net_Dataset,
                'optim' : torch.optim.Adam,
                'loss_func': hp.choice('loss_func', [nn.MSELoss(), nn.SmoothL1Loss()]),
                'num_epochs' : 4000,
                'learning_rate': hp.choice('learning_rate', [5e-3, 1e-2, 1e-1]),
                'batch_size' : hp.choice('batch_size', [64, 128, 192])
                }
```

```python
trials = Trials()
```

```python
results = fmin(
                objective,
                space=param_space,
                algo=tpe.suggest,
                trials=trials,
                max_evals=50,
                rstate=np.random.default_rng(RANDOM_STATE)
              )
```

```python
def getBestModelfromTrials(trials):
    valid_trial_list = [trial for trial in trials
                            if STATUS_OK == trial['result']['status']]
    losses = [ float(trial['result']['loss']) for trial in valid_trial_list]
    index_having_minumum_loss = np.argmin(losses)
    best_trial_obj = valid_trial_list[index_having_minumum_loss]
    return best_trial_obj['result']['model'], best_trial_obj['result']['params']

best_model, params_best = getBestModelfromTrials(trials)
```

```python
best_net = Net_Init(params_best['n_in_neurons'], 
           params_best['n_hidden_neurons_1'], 
           params_best['n_hidden_neurons_2'],
           params_best['n_out_neurons'],
           params_best['drop_1'],
           params_best['drop_2'],
           params_best['act_layer_1'],
           params_best['act_layer_2'],
           params_best['batch_norm'])

best_model = Net_Trainer(dataset=params_best['dataset'], 
                  net=best_net, 
                  optim=params_best['optim'], 
                  loss_func=params_best['loss_func'], 
                  num_epochs=params_best['num_epochs'], 
                  learning_rate=params_best['learning_rate'], 
                  batch_size=params_best['batch_size'])
```
